## Build: portfolio tracker

```python
holdings = {
    "AAPL": {"shares":20, "cost": 150.00, "price":187.46},
     "MSFT": {"shares": 10, "cost": 330.00, "price": 410.20},
    "NVDA": {"shares": 50, "cost": 135.00, "price": 120.55},
    "JPM":  {"shares": 15, "cost": 160.00, "price": 198.30},
}

print(holdings["NVDA"]["shares"])
```

```python
for ticker, data in holdings.items():
    shares = data["shares"]
    price = data["price"]
    cost = data["cost"]
    market_value = shares*price

    print(f"Market value of {ticker} is {market_value:,.2f}")

    
```

```python
nav = 0

for ticker, data in holdings.items():
    shares = data["shares"]
    price = data["price"]
    cost = data["cost"]
    market_value = shares*price
    nav += market_value

print(f"NAV: {nav:,.2f}")
```

```python
for ticker, data in holdings.items():
    shares = data["shares"]
    price = data["price"]
    cost = data["cost"]
    market_value = shares*price
    market_cost = shares*cost

    print(f"Market cost of {ticker} is {market_cost:,.2f}")
```

```python
c = 0

for ticker, data in holdings.items():
    shares = data["shares"]
    price = data["price"]
    cost = data["cost"]
    market_value = shares*price
    market_cost = shares*cost

    c += market_cost

print(f"Total portfolio cost is {c:,.2f}")
```

```python
Profit_Loss = nav - c

print(f"Profit&Loss is {Profit_Loss:,.2f}")
```

```python
percentage_profit = Profit_Loss / c

print(f"Percentage profit is {percentage_profit:,.2%}")
```

```python
for ticker, data in holdings.items():
    shares = data["shares"]
    price = data["price"]
    cost = data["cost"]
    market_value = shares*price
    market_cost = shares*cost
    weight = market_value / nav

    print(f"{ticker}'s weight is {weight:,.2%}")
```

```python
w = 0

for ticker, data in holdings.items():
    shares = data["shares"]
    price = data["price"]
    cost = data["cost"]
    market_value = shares*price
    market_cost = shares*cost
    weight = market_value / nav

    w += weight

print(f"Total weight is {w}")
```

## Challenge : rebalance

```python
desired_value = nav * 0.25


for ticker, data in holdings.items():
    shares = data["shares"]
    price = data["price"]
    cost = data["cost"]
    market_value = shares*price
    adjusted_amount = desired_value - market_value
    adjusted_percentage = adjusted_amount / nav

    print(f"(Need to adjust {ticker} by {adjusted_amount:,.2f}, so position is changed by {adjusted_percentage:,.2%})")

    

```

```python
for ticker, data in holdings.items():
    shares = data["shares"]
    price = data["price"]
    cost = data["cost"]
    market_value = shares*price
    adjusted_amount = desired_value - market_value
    adjusted_percentage = adjusted_amount / nav
    shares_to_trade = abs(adjusted_amount) / price

    if adjusted_amount < 0:
        print(f"Sell {shares_to_trade:,.2f} shares of {ticker}")
    elif adjusted_amount > 0:
        print(f"Buy {shares_to_trade:,.2f} shares of {ticker}")
    else:
        print(f"Don't change {ticker}'s position")
```
