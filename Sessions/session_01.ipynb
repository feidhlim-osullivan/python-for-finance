# Session 1: Variable, types, f-strings

```python
principal = 10000   #int: whole number
rate = 0.05         # float: decimal
years = 10          # int
ticker = "AAPL"     # str: text, always in quotes
is_annual = True    # bool: True or False

print(type(principal), type(rate), type(years), type(ticker), type(is_annual))
```

```python
fv = principal * (1 + rate) ** years
print(fv)
```

```python
print(f"{ticker} position grows to €{fv:,.2f} after {years} years")
print(f"Rate: {rate:.2%}")
```

```python
principal = 25000
rate = 0.045
years = 7

fv = principal * (1 + rate) ** years

print(f"{fv:,.2f}")

for n in [1, 4, 12, 365]:
    fv = principal*(1 + rate/n) ** (n * years)
    print(f"position grows to €{fv:,.2f} after {years} years")
        
```

```python
rate = 0.045
for n in [1, 4, 12, 365]:
    EAR = ((1 + rate/n) ** n) - 1
    print(f"EAR is {EAR:,.2%}")
```

```python
loan = 300000
annual_rate = 0.038
years = 30
pay_per_year = 12

monthly_rate = annual_rate / pay_per_year
n_payments = years * pay_per_year

PMT =  loan * monthly_rate * (1 + monthly_rate) ** n_payments / ((1 + monthly_rate) ** n_payments - 1)

print (f"Monthly payment is {PMT:,.2f}")
print(f"Total Paid over life of the loan is {PMT * n_payments:,.2f}")
print(f"Total Interest paid is {(PMT * n_payments - loan):,.2f}")
print(f"Percentage of interest over the loan is {(PMT * n_payments - loan)/loan:,.2%}")
```
